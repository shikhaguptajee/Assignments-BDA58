Data Cleaning

Proper handling of missing values, outliers, and data inconsistencies

```python
dataset['mainroad'] = dataset['mainroad'].map({'yes': 1, 'no': 0}) 
dataset['guestroom'] = dataset['guestroom'].map({'yes': 1, 'no': 0}) 
dataset['basement'] = dataset['basement'].map({'yes': 1, 'no': 0}) 
dataset['hotwaterheating'] = dataset['hotwaterheating'].map({'yes': 1, 'no': 0}) 
dataset['airconditioning'] = dataset['airconditioning'].map({'yes': 1, 'no': 0}) 
dataset['furnishingstatus'] = dataset['furnishingstatus'].map({'furnished': 2, 'semi-furnished': 1, 'unfurnished': 0})
dataset
```

```python
# Handle outliers, we are using capping with in the first and 99th percentile.
def cap_outliers(df, column): 
    lower_limit = df[column].quantile(0.01)
    upper_limit = df[column].quantile(0.99)
    df[column] = np.clip(df[column], lower_limit, upper_limit)

cap_outliers(dataset, 'price')
cap_outliers(dataset, 'area')
cap_outliers(dataset, 'bedrooms')
cap_outliers(dataset, 'bathrooms')
cap_outliers(dataset, 'stories')
cap_outliers(dataset, 'mainroad')
cap_outliers(dataset, 'guestroom')
cap_outliers(dataset, 'basement')
cap_outliers(dataset, 'hotwaterheating')
cap_outliers(dataset, 'airconditioning')
cap_outliers(dataset, 'parking')
cap_outliers(dataset, 'prefarea')
cap_outliers(dataset, 'furnishingstatus')
```
